## 15. Handling Missing Values

```python
#fill missing Salary_LPA with mean & median.
```

```python
JobMarket["Salary_LPA"] = JobMarket["Salary_LPA"].fillna(JobMarket["Salary_LPA"].mean())
JobMarket["Salary_LPA"].isnull().sum()
```

```python
JobMarket["Salary_LPA"] = JobMarket["Salary_LPA"].fillna(JobMarket["Salary_LPA"].median())
JobMarket["Salary_LPA"].isnull().sum()
```

```python
# fill Applicants with mode.
```

```python
JobMarket["Applicants"] = JobMarket["Applicants"].fillna(JobMarket["Applicants"].mode()[0])
JobMarket["Applicants"].isnull().sum()
```

```python
# fill forward.
```

```python
JobMarket.ffill(inplace=True)
```

```python
# backforward fill.
```

```python
JobMarket.bfill(inplace=True)
```

## 16. Checking Duplicates

```python
# checking duplicate rows.
```

```python
JobMarket.duplicated().sum()
```

## 17. Removing Duplicates

```python
JobMarket.drop_duplicates(inplace=True)
```

```python
#remove missing rows.
```

```python
JobMarket.dropna(inplace=True)
```

## 18. Rename Columns

```python
# renaming columns to improve readability and consistency in the dataset.
```

```python
JobMarket.rename(columns={
    "Job_Title": "Role",
    "Salary_LPA": "Salary",
    "Experience_Level": "Experience",
    "Required_Skills": "Skills"
}, inplace=True)
```

## 23. Convert Salary into Monthly Pay

```python
JobMarket["Monthly_Salary"] = (JobMarket["Salary"] * 100000) / 12
```

## 24. Salary Category

```python
JobMarket["salary_category"] = np.where(
    JobMarket["Salary"] >= 15,
    "High Salary",
    "Low Salary"
)
```

## 25. Applicant Category

```python
JobMarket["applicant_category"] = np.where(
    JobMarket["Applicants"] >= 45,
    "High Competition",
    "Low Competition"
)
```

## 26. Skill Count Feature

```python
JobMarket["Skill_Count"] = JobMarket["Skills"].apply(lambda x: len(str(x).split(",")))
```

## 27. Skill Count Category

```python
JobMarket["skill_count_category"] = np.where(
    JobMarket["Skill_Count"] >= 5,
    "Multi-Skill Role",
    "Focused Role"
)
```

## 28. Experience Category

```python
JobMarket["experience_category"] = np.where(
    JobMarket["Experience"].isin(["Senior (6-10 yrs)", "Lead (10+ yrs)"]),
    "Senior Level",
    "Entry/Mid Level"
)
```

# Final Business Insights

## Key Business Insights— 

### 1. Role Performance :

#### AI Research Engineer is the best-performing role in terms of average salary, while roles like Data Analyst and Software Engineer dominate in posting volume.

### 2. Skill Demand :

#### NLP, Java, Node.js and SQL are the most in-demand skills, showing strong overlap between AI/ML and traditional software-engineering capability.

### 3. Regional Performance :

#### Bengaluru generates the highest number of job postings among all locations, followed by Hyderabad and Pune.

### 4. Industry Performance :

#### IT Services leads in posting volume, while specialised industries such as FinTech offer stronger average salaries.

### 5. Hiring & Competition Behavior :

#### Hiring activity peaks mid-cycle across the year, and postings with higher applicant competition do not always correspond to the highest salaries.

# Conclusion ---->

This project analyzed the Job Market dataset to understand skill demand, salary benchmarks, hiring trends, role performance, industries, and
regional trends. Data cleaning, feature engineering, filtering, sorting, aggregation, statistical analysis, and visualization were performed to extract
meaningful insights. The analysis identified AI Research Engineer as the top-paying role, IT Services as the leading industry by posting volume, NLP as the
most in-demand skill, and Bengaluru as the top hiring location, with hiring activity trending upward and peaking mid-cycle through the year.
